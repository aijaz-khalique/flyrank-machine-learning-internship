## 1. My lane as an ML task (type)

My lane as an ML task

I am framing my lane as a **classification problem**.

The goal is to predict whether a content item is **declining or not declining** based on its available signals and characteristics.

This is a binary classification task because the output has two possible classes:

1 = Declining
0 = Not declining

The prediction can support editors or content teams by identifying items that may need attention or review.

```python
task_type = "Classification"

classes = ["Declining", "Not Declining"]

print("ML Task Type:", task_type)
print("Classes:", classes)
```

Target or proxy

The target I would predict is whether a content item is declining.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success metric

My main success metric is **Recall** for the declining class.

Recall measures how many of the truly declining content items the model successfully identifies.

This metric is important because the purpose of the model is to help identify content items that may need attention. Missing a declining item means it may not be reviewed.

A higher recall means the model finds more of the items that are actually declining.

I would consider the model useful if it achieves a recall that is meaningfully better than a simple baseline, while still maintaining reasonable precision so that editors are not given too many false alerts.

The model is intended as **decision-support**, not as an automatic decision-maker.

```python
metric = "Recall"

print("Primary success metric:", metric)
print("Goal: Identify as many truly declining content items as possible.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML beats a fixed rule here

A fixed rule might look like this:

If one metric is below a certain threshold, classify the content item as declining.

However, real content performance may depend on multiple signals interacting together. A content item may show a combination of engagement, traffic, historical, or other measured patterns that cannot be captured reliably by one simple threshold.

Machine learning can learn patterns across multiple variables at the same time and estimate the likelihood that a content item belongs to the declining class.

The purpose is not to replace human judgment. The model provides **decision-support** by helping prioritize which content items may deserve review.

A fixed rule uses manually chosen thresholds, while a machine learning model can learn relationships from the available data.

```python
print("Example of a fixed rule:")

print("""
IF one_metric < threshold:
    Predict declining
ELSE:
    Predict not declining
""")

print("A machine learning model can combine multiple signals instead of relying on one fixed threshold.")
```
